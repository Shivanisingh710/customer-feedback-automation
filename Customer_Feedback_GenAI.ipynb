## Filtering

#### Filter Reviews with Rating ≤ 2

```python
neg_df = df[df['Score']<=2].copy()
```

```python
print('Total reviews:', len(df))
print('Negative reviews:',len(neg_df), 'Total%:',(len(neg_df)/len(df))*100)
```

#### Filter Critical Keywords

```python
# I used Summary column to identify the most common complaint keywords because contains 
# concise description of customer's concerns that made easy to find most common complaint keywords.
```

```python
from collections import Counter

w = " ".join(neg_df['Summary']).split()
w1 = [i for i in w if len(i) > 3]
a = sorted(Counter(w1).most_common(60),key=lambda x: x[1],reverse=True)
a
```

```python
com_w = ["disappointed", "horrible", "terrible", "poor", "worst", "stale",
         "awful", "weak", "waste", "beware", "wrong", "bitter", "nasty","yuck"]

def top3_complaints(words, complaint_words):
    top3 = []
    for i,j in words:
        if i in complaint_words:
            top3.append((i, j))
    return top3[:3]
top3 = top3_complaints(a, com_w)
print(top3)
```

#### Select Top 3 Critical Reviews

```python
neg_df['Review_Length'] = neg_df['Text'].str.len()
filtered_reviews = neg_df[neg_df["Summary"].str.contains("|".join([i for i,j in top3]), case=False, na=False)]
top3_reviews = filtered_reviews.sort_values(by=["Score", "Review_Length"], ascending=[True, False])
```

```python
top3_reviews.head(3)
```

```python
# The three most critical reviews were selected by prioritizing lower ratings (1-star before 2-star). Among reviews with 
# the same rating, the longest reviews were chosen because they contain more detailed descriptions of customer issues, 
# enabling the Generative AI model to produce more personalized apology emails.
```

## Automated Outreach Using Generative AI

```python
from groq import Groq

client = Groq(api_key="Enter-Your-API-Key")
```

```python
emails = []

for i in range(3):
    profile_name = top3_reviews.iloc[i]["ProfileName"]
    review = top3_reviews.iloc[i]["Text"]

    prompt = f"""
You are a Customer Support Agent at Amazon.

Write a short, professional, and empathetic apology email (around 100 words) in response to the customer's review.

Customer Name: {profile_name}

Customer Review:
{review}

Instructions:
- Start the email with: Dear {profile_name},
- Acknowledge the customer's concerns.
- Apologize sincerely for their experience.
- Assure the customer that their feedback is valuable and the issue will be investigated.
- Maintain a polite and professional tone.
- Do not make unrealistic promises or offer compensation unless mentioned.
- End the email exactly as follows:

Sincerely,
Customer Support Team
Amazon
"""

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}]
    )

    emails.append(response.choices[0].message.content)
```

```python
for i, email in enumerate(emails, 1):
    print(f"\nEmail {i}\n")
    print(email)
    print("-" * 80)
```
